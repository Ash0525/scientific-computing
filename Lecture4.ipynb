```python
# break and continue
# find if it is prime or not
n = 2027


# m is the divisor
for m in range(2, n):
    if n%m ==0:
        print(f"{n} is not prime: {n} = {m} * {n//m}")
        break
    last = m

if last == n - 1:
    print(n, 'is a prime')
```
