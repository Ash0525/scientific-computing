```python
# Exercise 2.5 

# July 30, 2026

import numpy as np
import matplotlib.pyplot as plt

# Make f(x)
def f(x):
    return np.cbrt(x ** 5)

# Make T1(x)
def T1(x):
    return np.zeros_like(x, dtype=float)

x = 0

print(f(x))
print(T1(x))

# Use linspace to get the precision
x = np.linspace(-2, 2, 100)

plt.plot(x, f(x), label="f(x)")
plt.plot(x, T1(x), label="T1(x)")

plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
