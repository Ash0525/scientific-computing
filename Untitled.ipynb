```python
# linspace
# The last value is the number of entries. The last value is included. Generates 6 entries from 0 to 2.
# Equally spaced, divided by the number of entries minus 1.
v = np.linspace(0, 2, 6)
print(v)
```

```python
# arange
# The last element is the increment. It looks like the range function, but it's a vector.
u = np.arange(0, 1, 0.2)
print(u)
```

```python
# Use matplotlib
y1 = np.exp(x) - 1
y2 = x**2

# You put 101 because you have to do (2 - 0) / (101 - 1)
x = np.linspace(0, 2, 11)

# They need to have the same length
# k is black
# o is circle
plt.plot(x, y1, 'k-', label='$y=e^x-1$')
plt.plot(x, y2, 'bo--', label='$y=x^2$')

# Limits
plt.xlim(-1, 3)
plt.ylim(-1, 8)

# Labels
plt.xlabel('x')
plt.ylabel('y')
plt.title('Graphs of $y=e^x-1$ and $y=x^2$')
plt.legend()
plt.grid()

plt.show()
```

```python
# Practice problem
# Evaluate P(x)

def polynomial(c, x):
    p = 0
    
    for i in range(0, len(c)):
        p+=c[i]*x**i

    return p

c = [1,2,3]

print(polynomial(c, 100))
        
    
```

```python
# passing lists into functions

def swap(d, i, j):
    a = d[i]
    b = d[j]

    d[j] = a
    d[i] = b

swapped = [1,2,3,4]
print(swapped)

swap(swapped, 2, 3)

print(swapped)
```

```python
# Graph with matplotlib
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 1000)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1, '-k', label="$cos(x)$")
plt.plot(x, y2, '-b', label="$sin(x)$")
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
```
