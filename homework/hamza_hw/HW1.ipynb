# Problem B.13

```python
import numpy as np
import matplotlib.pyplot as plt

def convertTemp(isCtoF, temp):
    # If the boolean isCtoF is true, the function converts from Celsius to Fahrenheit
    # If false, the function converts from Fahrenheit to Celsius
    if isCtoF:
        return (temp * 9/5) + 32
    else:
        return (temp - 32) * 5/9

c = []
f = []
for i in range(-100,100):
    c.append(i)
    f.append(convertTemp(True,i))
plt.plot(c,f)
plt.title("Celsius to Fahrenheit")
plt.xlabel("Celsius")
plt.ylabel("Fahrenheit")
plt.show()

plt.plot(f,c)
plt.title("Fahrenheit to Celsius")
plt.xlabel("Fahrenheit")
plt.ylabel("Celsius")
plt.show()
```

# Problem B.14

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)
for i in range(0,11):
    print(str(i) + "! = " + str(factorial(i)))
```
