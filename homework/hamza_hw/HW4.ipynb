```python
# (1)
import numpy as np
import matplotlib.pyplot as plt

def nest(c,x,a):
    n = len(c)
    z = x-a
    p = c[n-1]
    num_add = 0
    num_mult = 0
    for i in range(n-2,-1,-1):
        p = c[i] + z * p
        num_add += 1
        num_mult += 1
#     print("Number of additions:",num_add);
#     print("Number of multiplications:",num_mult);
    return p

x = np.arange(-0.5,0.51,0.01)
f = np.log(1+x)
t_4 = nest([0,1,-1/2,1/3,-1/4],x,0)
t_9 = nest([0,1,-1/2,1/3,-1/4,1/5,-1/6,1/7,-1/8,1/9],x,0)
plt.plot(x,t_4,'k--',label='$T_4(x)$')
plt.plot(x,t_9,'r-',label='$T_9(x)$')
plt.plot(x,f,label='$\ln(1+x)$')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.title('Graphs of $\\ln(1+x)$, $T_4(x)$, and $T_9(x)$')
plt.show()
```

```python
# (2)
r_4 = f-t_4
r_9 = f-t_9
plt.semilogy(x,r_4,label='$R_4(x)$')
plt.semilogy(x,r_9,label='$R_9(x)$')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Logarithmic-scale y plot of error for $T_4(x)$ and $T_9(x)$')
plt.legend()
plt.show()
```

```python
# (3)
"""
The error of T_9(x) is less than the derived upper bound
"""
print("Maximum error for T_9(x):",np.max(r_9))
```

Using Taylor's theorem, we know, $|R_n(x)|=|\frac{f^{(n+1)}(c)}{(n+1)!}x^{n+1}|$

Since $-0.5\leq x\leq0.5$, $|x|\leq0.5$

We know the 10th derivative of f(x) is $-\frac{9!}{(x+1)^{10}}$

$0\leq c\leq 0.5$, so $|f^{(10)}(c)|\leq\frac{9!}{1^{10}}$

Thus, $|R_9(x)|\leq\frac{9!}{10!}0.5^{10}\approx9.76\times10^{-5}$
