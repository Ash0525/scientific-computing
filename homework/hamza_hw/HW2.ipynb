# Problem B.4

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2,2,21)
y1 = np.exp(x)-1
y2 = 1+x+np.power(x,2)/2+np.power(x,3)/6
plt.plot(x,y1,'k-',label='$y=e^x-1$')
plt.plot(x,y2,'bo--',label='$y=1+x+x^2/2+x^3/6$')
plt.xlabel('x')
plt.ylabel('y')
plt.title("Graph of $y=e^x-1$ and $y=1+x+x^2/2+x^3/6$")
plt.legend()
plt.show()
```
