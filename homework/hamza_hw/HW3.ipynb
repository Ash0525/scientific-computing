```python
def nest(c,x,a):
    n = len(c)
    z = x-a
    p = c[n-1]
    num_add = 0
    num_mult = 0
    for i in range(n-2,-1,-1):
        p = c[i] + z * p
        num_add += 1
        num_mult += 1
    print("Number of additions:",num_add);
    print("Number of multiplications:",num_mult);
    return p

# f(x)=1+3(x+1)-2(x+1)^3 centered at a=-1, evaluated at x=0
print("Result:",nest([1,3,0,-2],0,-1))
```
