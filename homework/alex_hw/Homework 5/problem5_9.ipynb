```python

import numpy as np
import matplotlib.pyplot as plt

m = 3
a = 2

x = np.linspace(-1, 2.5, 1001)
f = x**m - a

# Newton's method
xk = 1
fxk = xk**m - a
dfxk = m * xk**(m-1)
x_next = xk - fxk / dfxk

# Tangent line
tangent = fxk + dfxk * (x - xk)

# Plot
plt.figure(figsize=(9, 6))

plt.plot(x, f, label=r"$f(x)=x^3-2$", linewidth=2)
plt.plot(x, tangent, label=r"Tangent: $y=3x-4$",
         linewidth=2, linestyle="--")

# Mark points
plt.scatter(xk, fxk, color="black", zorder=5)
plt.scatter(x_next, 0, color="red", zorder=5)

# Annotate points
plt.annotate(r"$(x_k,f(x_k))=(1,-1)$",
             (xk, fxk), xytext=(-145, -25),
             textcoords="offset points")

plt.annotate(r"$x_{k+1}=\frac{4}{3}$",
             (x_next, 0), xytext=(15, -25),
             textcoords="offset points")

# Axes
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlim(-0.5, 2)
plt.ylim(-4, 5)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Newton's Method: Exercise 5.9 (3)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.show()
```
