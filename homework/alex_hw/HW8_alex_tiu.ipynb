# Problem 7.1

```python
# Trapezoidal Rule

def trapezoidal(f, a, b, n):
    """
    Accepts a function, f, the interval [a, b], and the number of rectangles, n
    """

    # Let h be the width of a rectangle
    h = (b - a) / n

    # Add the two end points
    Rt = (f(b) + f(a)) / 2

    # Iterate and sum to get all of the heights
    for k in range(1, n):

        # Let f(a + k * h) be the height of an interior rectangle
        Rt = Rt + f(a + k * h)

    return Rt * h
    
```

```python
# Implementation of the code
import numpy as np

# Use lambda function
f = lambda x: np.sin(np.pi * x)* np.exp(x)
partitions = [10, 20, 40, 80, 160, 320, 640, 1280]

a = -1
b = 1

# Let T be the answer to the integration
T = (np.pi * (np.e - (1 / np.e))) / (1 + (np.pi ** 2))

# For each partition calculate the h, value, error and ratio
# Note that n = 10 has no available ratio
Et0 = None

for n in partitions:

    Rt = trapezoidal(f, a, b, n)

    # Error
    Et = np.abs(Rt - T)
    
    # Width
    h = (b - a) / n

    # Skip 10 ratio
    if Et0 is None:
        print(f"h: {h} Rt: {Rt} Error: {Et}")
        Et0 = Et
    
    # Else print with ratio
    else:
        print(f"h: {h} Rt: {Rt} Error: {Et} Ratio: {Et0 / Et}")
```

```python
# Write Simpson's rule using pseudocode
def simpson(f, a, b, n):
    """
    Accepts function f, interval [a, b], and number of partitions, n
    """
    # Calculate widths
    h = (b - a) / n

    # Initialize S2. Function values that will receive weight of 2
    S2 = 0

    # Evaluate f at the required points (S2)
    for i in range(1, n):
        xi = a + i * h
        S2 += f(xi)
    
    # Initialize S4. Function values that will receive weight of 4
    S4 = 0

    # Evaluate f at the required points (S4)
    for i in range(1, n + 1):
        xm = a + i * h - h / 2
        S4 += f(xm)

    return (h / 6) * (f(a) + f(b) + 2 * S2 + 4 * S4)
```

```python
# Implementation of the code (same as 7.1)
import numpy as np

# Use lambda function
f = lambda x: np.sin(np.pi * x)* np.exp(x)
partitions = [10, 20, 40, 80, 160, 320]

a = -1
b = 1

# Let T be the answer to the integration
T = (np.pi * (np.e - (1 / np.e))) / (1 + (np.pi ** 2))

# For each partition calculate the h, value, error and ratio
# Note that n = 10 has no available ratio
Et0 = None

for n in partitions:

    Rt = simpson(f, a, b, n)

    # Error
    Et = np.abs(Rt - T)
    
    # Width
    h = (b - a) / n

    # Skip 10 ratio
    if Et0 is None:
        print(f"h: {h} Rt: {Rt} Error: {Et}")
        Et0 = Et
    
    # Else print with ratio
    else:
        print(f"h: {h} Rt: {Rt} Error: {Et} Ratio: {Et0 / Et}")
```
