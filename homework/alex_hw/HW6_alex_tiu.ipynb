```python
# Write the bisection method using the pseudocode

# Let a and b represent the endpoints. Let f be the function we are evaluating the roots for
def bisection_method(f, a, b, tolerance):

    # Counter
    k = 0

    # Plug in a and b into f
    fa = f(a)
    fb = f(b)

    # Get the midpoint c from a and b
    c = (a + b) / 2

    # If fa and fb have the same sign, the interval does not bracket a root.
    # We need an interval whose endpoint values have opposite signs (IVT).
    if (fa * fb > 0):
        raise ValueError(f"f(a)={fa} and f(b)={fb} must have different signs!")
    
    # While b - a stays above the tolerance
    while (b - a > tolerance):

        # Get the midpoint again
        c = (a + b) / 2

        # Find the value of the function at c
        fc = f(c)

        # If fc equals 0, then we have found the root!
        if fc == 0:
            return c
        
        # If there is a sign change in [a, c], keep that interval.
        elif fa * fc < 0:
            b = c
            fb = fc
        
        # Otherwise, the sign change is in [c, b].
        else:
            a = c
            fa = fc

        # Counter for number of iterations
        k += 1

    return (a + b) / 2, k
```

```python
# Tests
# Use 10 + x^3 - 12cos(x) = 0
import numpy as np

def f(x):
    return 10 + x**3 - 12 * np.cos(x)

e = 1e-4
a = -5
b = 5

root, iterations = bisection_method(f, a, b, e)

print(f"Let the interval be [{a},{b}]")
print(f"Let the tolerance e be: {e}")
print(f"For this interval to be a good interval to start with, one of the values should be negative")
print(f"f({a}) = {f(a)}")
print(f"f({b}) = {f(b)}")
print(f"Because one of them is negative, we can continue\n")
print("-" * 15)
print(f"Number of iterations: {iterations}")
print(f"Root is at: {root}")
print(f"f({root}) = {f(root)}")

```
