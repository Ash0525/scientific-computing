```python
# Trapezoidal Rule

def trapezoidal(f, a, b, n):
    """
    Accepts a function, f, the interval [a, b], and the number of rectangles, n
    """

    # Let h be the width of a rectangle
    h = (b - a) / n

    # Add the two end points
    Rt = (f(b) + f(a)) / 2

    # Iterate and sum to get all of the heights
    for k in range(1, n):

        # Let f(a + k * h) be the height of an interior rectangle
        Rt = Rt + f(a + k * h)

    return Rt * h
    
```

```python

# Implementation of the code
import numpy as np
import pandas as pd

# Use lambda function
f = lambda x: np.sin(np.pi * x) * np.exp(x)
partitions = [10, 20, 40, 80, 160, 320, 640, 1280]

a = -1
b = 1

# Let T be the answer to the integration
T = (np.pi * (np.e - (1 / np.e))) / (1 + (np.pi ** 2))

# For each partition calculate the h, value, error and ratio
# Note that n = 10 has no available ratio
Et0 = None
results = []

for n in partitions:

    Rt = trapezoidal(f, a, b, n)

    # Error
    Et = np.abs(Rt - T)

    # Width
    h = (b - a) / n

    # Skip 10 ratio
    if Et0 is None:
        ratio = ""

    # Else print with ratio
    else:
        ratio = Et0 / Et

    # Store the results in the table
    results.append({
        "n": n,
        "h": h,
        "Rt": Rt,
        "Error": Et,
        "Ratio": ratio
    })

    # Update the previous error
    Et0 = Et

# Display the results as a table
df = pd.DataFrame(results)
display(df)
```


### Discussion

As the number of subintervals increases, the composite trapezoidal approximation approaches the exact value of the definite integral. The absolute error decreases as the step size $h$ decreases. When $h$ is halved, the error decreases by approximately a factor of 4, as demonstrated by the error ratios approaching 4. This agrees with the theoretical error order of $O(h^2)$ for the composite trapezoidal rule.

```python
# Write Simpson's rule using pseudocode
def simpson(f, a, b, n):
    """
    Accepts function f, interval [a, b], and number of partitions, n
    """
    # Calculate widths
    h = (b - a) / n

    # Initialize S2. Function values that will receive weight of 2
    S2 = 0

    # Evaluate f at the required points (S2)
    for i in range(1, n):
        xi = a + i * h
        S2 += f(xi)
    
    # Initialize S4. Function values that will receive weight of 4
    S4 = 0

    # Evaluate f at the required points (S4)
    for i in range(1, n + 1):
        xm = a + i * h - h / 2
        S4 += f(xm)

    return (h / 6) * (f(a) + f(b) + 2 * S2 + 4 * S4)
```

```python

# Implementation of the code (same as 7.1)
import numpy as np
import pandas as pd

# Use lambda function
f = lambda x: np.sin(np.pi * x) * np.exp(x)
partitions = [10, 20, 40, 80, 160, 320]

a = -1
b = 1

# Let T be the answer to the integration
T = (np.pi * (np.e - (1 / np.e))) / (1 + (np.pi ** 2))

# For each partition calculate the h, value, error and ratio
# Note that n = 10 has no available ratio
Et0 = None
results = []

for n in partitions:

    Rs = simpson(f, a, b, n)

    # Error
    Es = np.abs(Rs - T)

    # Width
    h = (b - a) / n

    # Skip 10 ratio
    if Et0 is None:
        ratio = ""

    # Else print with ratio
    else:
        ratio = Et0 / Es

    # Store the results in the table
    results.append({
        "n": n,
        "h": h,
        "Rs": Rs,
        "Error": Es,
        "Ratio": ratio
    })

    # Update the previous error
    Et0 = Es

# Display the results as a table
df = pd.DataFrame(results)
display(df)
```


### Discussion

As the number of subintervals increases, the composite Simpson's rule approximation approaches the exact value of the definite integral. When the step size $h$ is halved, the absolute error decreases by approximately a factor of 16, as demonstrated by the error ratios approaching 16. This agrees with the theoretical error order of $O(h^4)$ for Simpson's rule. Compared with the composite trapezoidal rule, Simpson's rule produces significantly smaller errors for the same number of subintervals due to its higher order of accuracy.
