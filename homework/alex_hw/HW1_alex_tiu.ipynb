```python
# Loop to add all numbers between 1 and 99
sum = 0

# Range is from inclusive to exaclusive
for i in range(99 + 1):

    # If number is odd
    if (i % 2 != 0):
        sum += i

print(sum)
    
```

```python
# Problem B.6

# Use nested for loops to assign entires of a 5x5 matrix A

# make 5x5 matrix using lists
column = []

for i in range(1, 6):
    row = []
    for j in range(1, 6):
        row.append(i*j)
    
    column.append(row)

print(column)
```

```python
# Problem B.6 Alternative
import numpy as np

A = np.zeros((5,5))

for i in range(5):
    for j in range(5):
        A[i, j] = (i + 1) * (j + 1)
    
print(A)
```

```python
# Problem B.7 

# Use a for loop to satisfy the recurrence relation of the Fibonacci sequence, but do not use an array 
# to compute and display the first 20 terms

# n + 1, don't include 0
first = 1
second = 1

for i in range(20):
    print(first)

    # add up first and second bc fib
    next = first + second
    first = second
    second = next

# print last in sequence
print(second)
```

```python
# Problem B.8

# If bank account balance is less than 5000, the annual interest rate is 1%. 
# If the account balance is betwee 5000 and 20,000 exclusive, the interest rate is 2%
# Otherwise, it's 5%
accounts = [1000,2000,55000,100287,50]

def rates(x):
    if (x < 5000):
        print(f"1% rate: {x * 1.01}")
    elif (x < 20000):
        print(f"2% rate: {x * 1.02}")
    else:
        print(f"5% rate: {x * 1.05}")

for price in accounts:
    rates(price)
```

```python
# Problem B.9

# Write a function that takes as input three variables and returns the largest of the three. 
# Do not use max()

# Step 1: begin function
def get_max(x, y, z):
    if (x >= y) and (x >= z):
        return x
    elif (y >= x) and (y >= z):
        return y
    else:
        return z
    
print(get_max(7,8,5))
print(get_max(70,10,1))
print(get_max(-1,20,25))
```

```python
# Problem B.10

# Let the variable epsilon equal to 1. Use a while loop to keep dividing epsilon by 2 until 
# epsilon < 10^-6, and determine how many divisions were used.

# Create a counter for how many divisions are used
divisions = 0

# Initialize epsilon and tolerance
epsilon = 1
TOL = 1e-6

while (epsilon > TOL):
    epsilon /= 2
    divisions += 1

print(f"Num of divisions: {divisions}")
```

```python
# Problem B.13

# Write a function which converts temperature between Celsius (C) and Fahrenheit (F)

import numpy as np
import matplotlib.pyplot as plt
# Write function for celsius to fahrenheit
def ctof(temp):
    return ((9 / 5) * temp) + 32

def ftoc(temp):
    return (5 / 9) * (temp - 32)

# Build the graphs
# Have a linspace
x = np.linspace(0, 100, 1000)
y = ctof(x)

plt.plot(x, y)
plt.xlabel("Celsius ()")
plt.ylabel("Fahrenheit")
plt.title("Fahrenheit vs. Celsius")
```

```python
# Build the graphs
# Have a linspace
x = np.linspace(0, 100, 1000)
y = ftoc(x)

plt.plot(x, y)
plt.xlabel("Fahrenheit ()")
plt.ylabel("Celsius")
plt.title("Celsius vs. Fahrenheit")
```
