```python
# 2.5 Chapter 2 Programming Problems

# Let c be the coefficients of the polynomial of degree n (list)
# Let x be the point that we want to evaluate at
# Let a be the shift

def nest(c, x, a):

    # Initialize result
    result = 0
    sum = 0
    multiply = 0
    offset = x - a

    # Iterate through the list of coefficients and apply the Horner algorithm
    for coefficient in reversed(c):

        # number of sums and multiplications
        if (coefficient != 0):
            sum += 1
            multiply += 1

        # Let the result be the result times the number plus the coefficient
        result = result * offset + coefficient

    print(f"Number of additions: {sum}")
    print(f"Number of multiplications: {multiply}")

    return result

# Test function p(x) = 1 + 3 * (x + 1) - 2 * (x + 1) ^ 3
c = [1, 3, 0, -2]
x = 0
a = -1

print(nest(c, x, a))
```
