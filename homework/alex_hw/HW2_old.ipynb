## Part 1

```python
# Make a matrix A and vector b

import numpy as np

A = np.array([
    [0,-6,8,1],
    [-2,5,5,-3],
    [7,8,0,3]
])

b = np.array([3,2,-13])

# Matrix addition: +
# Element multiplicaiton: *
# Matrix multiplication: @

print(A)
print(b)
```

## Part 2

```python
# Find the sizes of A and b
print(f"Size of A: {A.size}")
print(f"Size of b: {b.size}")
```

## Part 3

```python
# Find the entry of A at row 2 and column 3 and write a command to get the third entry in b

print(f"Entry at row 2 column 3 of A: {A[1, 2]}")
print(f"Third entry of b: {b[2]}")
```

## Part 4

```python
# Swap second and third rows
# A[[row1, row2]] = A[[row2, row1]]

A[[1,2]] = A[[2,1]]

print(f"Swapped second and third rows:\n{A}")
```

# Problem B.2

```python
# Write a numpy program to get the powers of array values element-wise

# Make a numpy array
a = np.array([1,2,3,4,5])

# Iterate through, use list comprehension
powers = [num**2 for num in a]

# Use numpy
b = 2
powers_np = np.power(a,  b)

# Display powers
print(f"Example list: {a}")
print(f"Using list comprension: {powers}")
print(f"Using numpy power function: {powers_np}")
```

# Problem B.3

```python
# Generate a row vector with entries that are even numbers between 10 and 0

# create vector with zero padding
c = np.zeros(6, dtype=int)

# from 0 to 10, splice and get even numbers
for index, value in enumerate(range(0, 11, 2)):
    c[index] = value

# Check if it works
print(f"Even list: {c}")

# Reverse the order. To reverse, make the splicer negative, no beginning and ending
c_reverse = c[::-1]

# Print reversed
print(f"Reversed even list: {c_reverse}")
```

```python
# another way to solve B.3

# Make a variable array
v = np.array([])

# Loop through starting at 10 to -2, exclusive
for i in range(10, -2, -2):
    v = np.append(v, i)

# Print
print(f"Reversed using another method: {v}")
```

# Problem B.4

```python
# Plot the graphs of functions y = e^x and
# y = 1 + x + x^2/2 + x^3/6
# Use 21 data points with equally spaced values

import matplotlib.pyplot as plt

# Make linspace
x = np.linspace(-2, 2, 21)

# Get the functions
def e(x):
    return np.exp(x)

def y(x):
    return 1 + x + (x**2 / 2) + (x**3 / 6)

# Graph the functions
plt.plot(x, e(x), 'bo-', label="e^x")
plt.plot(x, y(x), 'ro-', label="1 + x + x^2/2 + x^3/6")

plt.title("Problem B.4")
plt.ylabel("y-axis")
plt.xlabel("x-axis")
plt.legend(loc="upper left")
plt.show()
```
