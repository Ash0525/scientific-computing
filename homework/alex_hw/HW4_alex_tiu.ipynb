```python
# Let Tn(x) bet the taylor polynomial of degree n for f(x) = ln(1 + x) at a = 0

import numpy as np

# Get the nest function from the other file

def nest(c, x, a):

    # Initialize result
    result = 0
    offset = x - a

    # Iterate through the list of coefficients and apply the Horner algorithm
    for coefficient in reversed(c):

        # Let the result be the result times the number plus the coefficient
        result = result * offset + coefficient

    return result
```

## Part 1

```python
# 1) Plot f(x) for T4(x) and T9(x)

import matplotlib.pyplot as plt

# Let x be the linspace
x = np.linspace(-0.5,0.5,101)

# Get coefficients for T4 and T9
c4 = [0, 1, -1/2, 1/3, -1/4]

c9 = c4 + [1/5, -1/6, 1/7, -1/8, 1/9]

f = np.log(x + 1)

t4 = nest(c4, x, 0)
t9 = nest(c9, x, 0)

plt.plot(x, t4, 'b-.', label="t4")
plt.plot(x, t9, 'r-', label="t9")
plt.plot(x, f, 'k--', label="f(x) = ln(x + 1)")
plt.legend()
plt.title("f(x), T4, T9")
plt.grid(True)
plt.ylabel("y")
plt.xlabel("x")
plt.show()
```

## Part 2

```python
# 2) Plot the Taylor's remainders (the error terms) |f(x)-T4(x)| and |f(x)-T9(x)|

err4 = f - t4
err9 = f - t9

plt.semilogy(x, err4, 'b--', label='$R_4(x)$')
plt.semilogy(x, err9, 'r-', label='$R_9(x)$')

plt.xlabel('x')
plt.ylabel('Absolute error')
plt.legend()
plt.grid(True)
plt.show()
```

## Part 3

### Taylor's Theorem Error Bound

For $T_9(x)$, we use the 10th derivative:

$$
\left|f^{(10)}(c)\right|
=
\frac{9!}{(1+c)^{10}}.
$$

To find the largest worst-case upper bound on $[-0.5,0.5]$, plug in $c=-0.5$:

$$
\left|f^{(10)}(-0.5)\right|
=
\frac{9!}{(0.5)^{10}}.
$$

Thus,

$$
\left|R_9(x)\right|
\leq
\frac{9!}{10!(0.5)^{10}}(0.5)^{10}
=
0.1.
$$
