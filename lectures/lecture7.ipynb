# Errors

Absolute and Relative Errors

T - true value of a quantity
A - approximation of true value (T)

Now we can define errors

We can find the difference between the true value and the approximate value. We want to reduce error for good approximation.
E - error
R - relative error

Error:
$E = T - A$

Absolute Error:
$|E| = |T - A|$

Relative Error:
$R = \frac{|E|}{|T|} = \frac{|T - A|}{|T|}$

True errors are unknown. So we need upper bound.


# Upper Bounds
U - Upper bound

$|E| = |T - A| \leq U$
