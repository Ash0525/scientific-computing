```python
# September 2, 2026
import math

# Implement a Python function
def quadroots(a, b, c):
    """
    This function finds the roots of a quadratic equation.
    Assume that b^2-4ax = 0, and use the quadratic formula.
    Inputs: coefficients a, b, c
    Outputs: roots x1 and x2
    """
    if a == 0:
        raiseValueError("a cannot be 0")

    discriminant = math.sqrt(b**2 - 4 * a * c)

    if discriminant < 0:
        raiseValueError("This function only solves for positive discriminant")
        
    x1 = (-b + discriminant) / (2 * a)
    x2 = (-b - discriminant) / (2 * a)

    return x1, x2

x1, x2 = quadroots(1, -5, 6)

print(x1, x2)

# You cannot access the variables (arguments) in the function because they are private
# They are positional
```

```python
# Another example
def my_power(base, exponent):
    """
    This function computes exponents
    Inputs: base, exponent
    Outputs: solution
    """
    return base ** exponent

solution = my_power(2, 3)

print(solution)
```

```python
# Example with passing list
def change_a_list(alist):
    alist[0]=-1
    alist.append(1)

# define the list
list_sent=[0]

# The function will change the list
print(f"Before calling function: {list_sent}")
change_a_list(list_sent)
print(f"After calling function: {list_sent}")
```

```python
# Example with adding
def addone(a):
    a+=1

# The variable in the function dies after addone. it was not stored anywhere public
a = 2
print(a)
addone(a)
print(a)
```

```python
# another practice
# Swap positions of list

# This is with swapping trick
def swap(list1, index1, index2):
    list1[index1], list1[index2] = list1[index2], list1[index1]

example = [1,2]
print(f"Before swap: {example}")
swap(example, 0,1)
print(f"After swap: {example}")

def naive_swap(list1, i1, i2):
    tmp = list1[i1]
    list1[i1] = list1[i2]
    list1[i2] = tmp

mylist=[0,1,2,3,4]
naive_swap(mylist,1,2)
print(mylist)
    
```
