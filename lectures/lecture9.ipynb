Polynomials are defined as:

$p_0(x) = c_0 + c_1(x-a) + c_2(x-a)^2 + \ldots + c_n(x-a)^n$

degree: the highest exponent.

We can use multiplication and addition to do this. This has to be done because things like sine and cosine cannot be calculated with a CPU. Polynomials can be easily evaluated with simple arithmetic.
