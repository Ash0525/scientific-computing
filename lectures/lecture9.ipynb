Polynomials are defined as:

$p_n(x) = c_0 + c_1(x-a) + c_2(x-a)^2 + \ldots + c_n(x-a)^n = \sum_{K=0}^{n} C_k(x-a)^k$

degree: the highest exponent.

We can use multiplication and addition to do this. This has to be done because things like sine and cosine cannot be calculated with a CPU. Polynomials can be easily evaluated with simple arithmetic. Also is easy to differentiate and integrate.

If a = 0,
then we get a special case

$p_n(x) = c_0 + c_1x + c_2x^2 + \ldots + c_nx^n$

## Examples

Zero polynomial: $z(x) = 0$
- The zero polynomial has an undefined degree

Linear polynomials have degree of one
Quadratic polynomials have degree of two

Two methods of Polynomial Evaluation

Example 1.1 Evaluate $p_4(x) = 1 + 2x + 3x^2 + 4x^3 + 5x^4$ at $x = 6$

Naive method:
$ 2x = (2)(6) \\
3x^2 = (2)(6)(6) \\
4x^3 = (2)(6)(6)(6) \\
5x^4 = (5)(6)(6)(6)(6)$

Number of total multiplications: 10

Number of total additions: 4

The total cost is 14

Generalization to a degree n polynomial: 
- Total number of multiplications: $1 + 2 + 3 + \ldots + n = \frac{(n+1)n}{2}$
- Total number of additions: $n$



You can make this faster by using Horner's method.
- We will use this method and implement it in Python

We can rewrite $p_4(x)$ in nested form:

$= 1 + x(2 + 3x + 4x^2 + 5x^3)\\
=1 + x(2 + x(3 + x(4 + 5x)))$

Number of multiplications: 4

Number of additions: 4

The total cost is 8

Generalization to a degree n polynomial:
- Total number of multiplications: $n$
- Total number of additions: $n$
