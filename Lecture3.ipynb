```python
# Lecture 3

# Exercise of an alternating series
stuff = []
for i in range(1, 1015):
    thing = ((-1)**(i+1)*(1/(2*i-1)))
    stuff.append(thing)

doodoo = sum(stuff)
print(4*doodoo)
        
```

```python
# Nested for loops and nested lists

# List with a list in it, 2x3
A=[[0,0,0],[0,0,0]]

for i in range(len(A)):
    for j in range(len(A[0])):
        
        # [0 0 0]
        # [0 0 0]
        # ith entry and jth entry
        A[i][j] = i * j

print(A)
```
