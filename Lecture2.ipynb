# Lists

```python
my_list = ['item 0', 1, 2.0, 'end']
# You can have a bunch of stuff in it, different data types

for i in my_list:
    print(i)
```

```python
# Accessing items in the list. Using index
first_item = my_list[0]
third_item = my_list[2]

last_item = my_list[-1]

print(first_item)
print(third_item)
print(last_item)
```

```python
# Slicing a list. Access a chunk of items. Specify a range of indices
# The last one is not included. It is exclusive. Inclusive to exclusive.
first_slice = my_list[1:3]

# Get to the end
to_end = my_list[2:]

# From beginning
from_start = my_list[:3]

print(first_slice)
print(to_end)
print(from_start)
```

```python
# Methods
# Length
print(len(my_list))
```

```python
# Append, add stuff
my_list.append('true end')

print(my_list)
```

```python
my_list[3] = 'second last'
print(my_list)
```

```python
# Tuple, immutable data type
# use it if you want to not change your dimensions
my_tuple = (1, 2, 3)
```

```python
# Range function
# Default increment is 1
# Range is exclusive 
list_by_range = list(range(-2,4))
print(list_by_range)

# Third number is the increment
list_by_range = list(range(4, -5, -3))
print(list_by_range)
```

```python
# For loop
# i is the loop variable
for i in my_list:
    print(i)
    
```

```python
# Example
cubes = []

for num in range(1, 11):
    cube = num**3
    cubes.append(cube)

print(cubes)
```

```python
s = 0
for k in range(1, 101):
    s += k
print(s)
```
