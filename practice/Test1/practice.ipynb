```python
# Make list version of the Fibonacci sequence. start from anywhere given a pair of fibonacci numbers

def fib(a, b, num):
    """
    Where a is the starting fib number, b is the next fib number, and num is the range for which we go.
    Note that this gives the number of iterations that you put into the function because we first append
    the first fib number
    """

    fib_list = []

    fib_list.append(a)
    for i in range(num-1):
        c = a + b
        a = b
        b = c
        fib_list.append(a)

    return fib_list

fib_list = fib(1, 1, 20)
for i in fib_list:
    print(i)
```

```python
# Recursion with the fibonacci sequence

def Fib(a):
    if a == 0:
        return 0
    elif a == 1:
        return 1
    else:
        return Fib(a-1) + Fib(a-2)


print(Fib(4))
```
