# Problem B.8

```python
# Bank accounts
accounts = [1000,2000,55000,100287,50]

with_interest = []

for account in accounts:
    if account < 5000:
        interest = 1.01 * account
        with_interest.append(interest)
    elif account < 20000:
        interest = 1.02 * account
        with_interest.append(interest)
    else:
        interest = 1.05 * account 
        with_interest.append(interest)

print(f"Original accounts: {accounts}")
print(f"Accounts with interest: {with_interest}")
```

# Problem B.10

```python
# Keep dividing epsilon
e = 1
num_divisions = 0

while e >= 10**-6:
    e /= 2
    num_divisions += 1

print(num_divisions)
```

# Problem B.12

```python
# Write a function that will convert radians into degrees
import numpy as np

def rad_to_deg(rad):
    deg =rad * (180 / np.pi)
    return deg
print(rad_to_deg(np.pi / 6))
print(rad_to_deg(np.pi / 4))
print(rad_to_deg(np.pi / 3))
print(rad_to_deg(np.pi / 2))
print(rad_to_deg(np.pi))
```

# Problem B.13

```python
# Celsius to fahrenheit
def c_to_f(c):
    return (9 / 5) * c + 32

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 100)
y = c_to_f(x)

plt.plot(x, y, '-o')
plt.xlabel('Celsius')
plt.ylabel('Fahrenheit')
```

# Problem B.14

```python
# recursive function for factorials

def factorial(n):
    # zero and 1 are weird cases
    if n == 0 or n == 1:
        return n
    else:
        return n * factorial(n-1)

for i in range(11):
    print(factorial(i))
```
