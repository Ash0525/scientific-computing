```python
# September 2, 2026
import math

# Implement a Python function
def quadroots(a, b, c):
    """
    This function finds the roots of a quadratic equation.
    Assume that b^2-4ax = 0, and use the quadratic formula.
    Inputs: coefficients a, b, c
    Outputs: roots x1 and x2
    """
    if a == 0:
        raiseValueError("a cannot be 0")

    discriminant = math.sqrt(b**2 - 4 * a * c)

    if discriminant < 0:
        raiseValueError("This function only solves for positive discriminant")
        
    x1 = (-b + discriminant) / (2 * a)
    x2 = (-b - discriminant) / (2 * a)

    return x1, x2

x1, x2 = quadroots(1, -5, 6)

print(x1, x2)

# You cannot access the variables (arguments) in the function because they are private
# They are positional
```

```python
# Another example
def my_power(base, exponent):
    """
    This function computes exponents
    Inputs: base, exponent
    Outputs: solution
    """
    return base ** exponent

solution = my_power(2, 3)

print(solution)
```
