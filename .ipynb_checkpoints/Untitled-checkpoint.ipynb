```python
# Modules are toolboxes. You can share your toolbox with other people
# math, numpy, matplotlib, scipy (useful for scientific computing)
# tensorflow, pytorch, keras, pandas, opencv (machine learning)

# Plotting
import matplotlib.pyplot as plt

# Numerical Python
import numpy as np

# Math
import math
```

```python
x = math.cos(math.pi)
print(x)
```

```python
# exponential
print(math.exp(1))
```

```python
# natural logarithm
print(math.log(1))
```

```python
# linspace
# The last value is the number of entries. The last value is included. Generates 6 entries from 0 to 2.
# Equally spaced, divided by the number of entries minus 1.
v = np.linspace(0, 2, 6)
print(v)
```

```python
# arange
# The last element is the increment. It looks like the range function, but it's a vector.
u = np.arange(0, 1, 0.2)
print(u)
```

```python
# Use matplotlib
y1 = np.exp(x) - 1
y2 = x**2

# You put 101 because you have to do (2 - 0) / (101 - 1)
x = np.linspace(0, 2, 101)

# They need to have the same length
plt.plot(x, y1, '-o')
plt.plot(x, y2, '-o')
```
